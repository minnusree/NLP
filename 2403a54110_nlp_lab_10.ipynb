```python
# Loading pre-trained embeddings
import gensim.downloader as api

# Handling matrices and arrays
import numpy as np

# t-SNE dimensionality reduction
from sklearn.manifold import TSNE

# Visualization
import matplotlib.pyplot as plt
```

```python
import gensim.downloader as api

# Load pre-trained Word2Vec model (300 dimensional)
model = api.load("word2vec-google-news-300")

# Print vocabulary size
print("Vocabulary Size:", len(model.key_to_index))

# Display one example vector
example_word = "computer"

print("Vector for 'computer':")
print(model[example_word])
print("Vector dimension:", len(model[example_word]))
```

```python
words = [

    # Animals
    "dog", "cat", "lion", "tiger", "elephant", "horse", "wolf", "bear",

    # Countries
    "India", "China", "France", "Germany", "Japan", "Brazil", "Canada", "Australia",

    # Fruits
    "apple", "banana", "mango", "orange", "grape", "pineapple", "pear", "peach",

    # Technology
    "computer", "laptop", "keyboard", "software", "internet", "mobile", "tablet", "processor",

    # Vehicles
    "car", "bus", "train", "airplane", "bicycle", "truck", "motorcycle", "ship"
]
```

```python
vectors = np.array([model[word] for word in words])
print("Vector matrix shape:", vectors.shape)
```

```python
tsne = TSNE(n_components=2, random_state=42, perplexity=5)

reduced_vectors = tsne.fit_transform(vectors)

print("Reduced shape:", reduced_vectors.shape)
```

```python
plt.figure(figsize=(12, 8))

x = reduced_vectors[:, 0]
y = reduced_vectors[:, 1]

plt.scatter(x, y)

for i, word in enumerate(words):
    plt.annotate(word, (x[i], y[i]))

plt.title("t-SNE Visualization of Word2Vec Embeddings")
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.grid(True)
plt.show()
```
